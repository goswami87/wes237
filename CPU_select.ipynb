```python
import sys
import os

user_input = int(input("Please enter CPU to use, either 0 or 1: "))
nTerms = int(input("enter fibonacci terms: "))

#print(user_input)

if user_input >= 2:
    sys.exit("only 0 or 1 CPU ID is allowed")

if nTerms <= 1:
    sys.exit("enter fibonacci term greater than 1")

commandToExecute = "taskset -c {} python3 fib.py {}".format(user_input,nTerms)

print("executing command on CPU {}: {}".format(user_input,commandToExecute))

os.system(commandToExecute) 
```
