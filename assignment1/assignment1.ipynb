```python
from pynq.overlays.base import BaseOverlay
import pynq.lib.rgbled as rgbled 
import time

base = BaseOverlay("base.bit")

btns = base.btns_gpio
led4 = rgbled.RGBLED(4)
led5 = rgbled.RGBLED(5)
led6 = rgbled.RGBLED(6)

#for eye to notice the LED must change every 10ms, i.e. frequency is 1000/10=100Hz

#frequency = 100 #in Hz
#dutyCycle = 0.25 # for example if one cycle is of 1Sec, it will stay on for 0.25sec and off for 0.75sec
#oneCycleLength=1/frequency

#onTime = dutyCycle*oneCycleLength
#offTime = (1-dutyCycle)*oneCycleLength

#At this ON/OFF rate, LED appears to be flickering
'''
while True:
    led4.write(0x7)
    time.sleep(0.025)
    led4.write(0x0)
    time.sleep(0.075)
    if btns.read() != 0:
        break

led4.write(0x0)
'''

#At this ON/OFF rate, LED doesn't appear to be flickering at all
'''
while True:
    led4.write(0x7)
    time.sleep(0.0025)
    led4.write(0x0)
    time.sleep(0.0075)
    if btns.read() != 0:
        break

led4.write(0x0)
'''

#75% on, more bright compared to previous one
'''
while True:
    led4.write(0x7)
    time.sleep(0.0075)
    led4.write(0x0)
    time.sleep(0.0025)
    if btns.read() != 0:
        break

led4.write(0x0)
'''

#time to find lowest possible frequency

# behavior is same as 100Hz, 50Hz i.e. no flickering observed
# Flickering observed with 35Hz, 25Hz, 10Hz 

frequency = 45 #in Hz
dutyCycle = 0.25 # for example if one cycle is of 1Sec, it will stay on for 0.25sec and off for 0.75sec
oneCycleLength=1/frequency

onTime = dutyCycle*oneCycleLength
offTime = (1-dutyCycle)*oneCycleLength

while True:
    led4.write(0x7)
    time.sleep(onTime)
    led4.write(0x0)
    time.sleep(offTime)
    if btns.read() != 0:
        break

led4.write(0x0)
```
